```python
#Open cv code
import cv2
import matplotlib.pyplot as plt
img = cv2.imread(r"C:\Users\marun\Downloads\ccat.gif")
rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
hsv = cv2.cvtColor(img, cv2.COLOR_RGB2HSV)
plt.subplot(1,2,1)
plt.imshow(rgb)
plt.title("Original Image")
plt.subplot(1,2,2)
plt.imshow(hsv)
plt.title("HSV Image")
plt.axis("off")
plt.show()
```

```python
#scikit image code 
from skimage import io, color
import matplotlib.pyplot as plt
img = io.imread(r"C:\Users\marun\Downloads\dog.jpg")
hsv = color.rgb2hsv(img)
plt.subplot(1,2,1)
plt.imshow(img)
plt.title("Original RGB Image")
plt.subplot(1,2,2)
plt.imshow(hsv)
plt.title("HSV Image")
plt.axis("off")
plt.show()
```
