```python
import cv2
import matplotlib.pyplot as plt

# Read image
img = cv2.imread(r"C:\Users\marun\Downloads\350788258498163951.jpg")
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    # Applying filters
median_blur = cv2.medianBlur(img, 9)
gaussian_blur = cv2.GaussianBlur(img, (9, 9), 10)
bilateral_blur = cv2.bilateralFilter(img, 9, 100, 75)

    # Convert outputs to RGB
median_rgb = cv2.cvtColor(median_blur, cv2.COLOR_BGR2RGB)
gaussian_rgb = cv2.cvtColor(gaussian_blur, cv2.COLOR_BGR2RGB)
bilateral_rgb = cv2.cvtColor(bilateral_blur, cv2.COLOR_BGR2RGB)

    # Displaying images
plt.figure(figsize=(12, 8))

plt.subplot(2, 2, 1)
plt.imshow(img_rgb)
plt.title("Original Image")
plt.axis("off")

plt.subplot(2, 2, 2)
plt.imshow(median_rgb)
plt.title("Median Blur")
plt.axis("off")

plt.subplot(2, 2, 3)
plt.imshow(gaussian_rgb)
plt.title("Gaussian Blur")
plt.axis("off")

plt.subplot(2, 2, 4)
plt.imshow(bilateral_rgb)
plt.title("Bilateral Filter")
plt.axis("off")
plt.tight_layout()

plt.show()
```
