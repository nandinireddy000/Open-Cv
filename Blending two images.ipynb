```python
import cv2
import matplotlib.pyplot as plt

# Reading the two images
img1 = cv2.imread(r"C:\Users\marun\Downloads\cat1.jpg")
img2 = cv2.imread(r"C:\Users\marun\Downloads\cat2.jpg")

# Resizing second image to match first image size
img2 = cv2.resize(img2, (img1.shape[1], img1.shape[0]))

# Blending the images
blended = cv2.addWeighted(img1, 0.5, img2, 0.5, 0)

# Convert BGR to RGB for display
img1_rgb = cv2.cvtColor(img1, cv2.COLOR_BGR2RGB)
img2_rgb = cv2.cvtColor(img2, cv2.COLOR_BGR2RGB)
blended_rgb = cv2.cvtColor(blended, cv2.COLOR_BGR2RGB)

# Display images
plt.figure(figsize=(12,4))

plt.subplot(1,3,1)
plt.imshow(img1_rgb)
plt.title("Image 1")
plt.axis("off")

plt.subplot(1,3,2)
plt.imshow(img2_rgb)
plt.title("Image 2")
plt.axis("off")

plt.subplot(1,3,3)
plt.imshow(blended_rgb)
plt.title("Blended Image")
plt.axis("off")

plt.show()
```
