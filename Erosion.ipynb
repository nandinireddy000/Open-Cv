```python
import cv2
import matplotlib.pyplot as plt

# Read image
image = cv2.imread(r"C:\Users\marun\Downloads\350788258498163951.jpg")

# Convert BGR to RGB
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

# Kernel sizes
s1 = 0
s2 = 10
s3 = 10

# Kernel types
t1 = cv2.MORPH_RECT
t2 = cv2.MORPH_CROSS
t3 = cv2.MORPH_ELLIPSE

# Create kernels
kernel1 = cv2.getStructuringElement(t1, (2*s1+1, 2*s1+1))
kernel2 = cv2.getStructuringElement(t2, (2*s2+1, 2*s2+1))
kernel3 = cv2.getStructuringElement(t3, (2*s3+1, 2*s3+1))

# Apply erosion
erode1 = cv2.erode(image, kernel1)
erode2 = cv2.erode(image, kernel2)
erode3 = cv2.erode(image, kernel3)

# Convert to RGB
erode1 = cv2.cvtColor(erode1, cv2.COLOR_BGR2RGB)
erode2 = cv2.cvtColor(erode2, cv2.COLOR_BGR2RGB)
erode3 = cv2.cvtColor(erode3, cv2.COLOR_BGR2RGB)

# Display results
plt.figure(figsize=(12,8))

plt.subplot(2,2,1)
plt.imshow(image_rgb)
plt.title("Original")
plt.axis("off")

plt.subplot(2,2,2)
plt.imshow(erode1)
plt.title("Erosion - Rectangle")
plt.axis("off")

plt.subplot(2,2,3)
plt.imshow(erode2)
plt.title("Erosion - Cross")
plt.axis("off")

plt.subplot(2,2,4)
plt.imshow(erode3)
plt.title("Erosion - Ellipse")
plt.axis("off")

plt.tight_layout()
plt.show()
```
