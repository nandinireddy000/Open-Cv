```python
import cv2
import matplotlib.pyplot as plt

# Read image
img = cv2.imread(r"C:\Users\marun\Downloads\cat1.jpg")

# Adding text
cv2.putText(
    img,
    "Hi There",      # Text
    (40, 95),           # Position (x, y)
    cv2.FONT_HERSHEY_SIMPLEX,  # Font
    2,                   # Font scale
    (0, 255, 0),         # Color (B, G, R)
    3,                   # Thickness
    cv2.LINE_AA          # Line type
)

# Convert BGR to RGB
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Display image
plt.imshow(img_rgb)
plt.title("Image with Text")
plt.axis("off")
plt.show()
```
