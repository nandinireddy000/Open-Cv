```python
#open cv code
import cv2
import matplotlib.pyplot as plt

# Read image
img = cv2.imread(r"C:\Users\marun\Downloads\dog.jpg")

# Convert BGR to RGB
rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Convert RGB to HSV
hsv = cv2.cvtColor(rgb, cv2.COLOR_RGB2HSV)

# Display
plt.imshow(hsv)
plt.title("HSV Image")
plt.axis("off")
plt.show()
```

```python
#scikit code
from skimage import io, color
import matplotlib.pyplot as plt

# Read image
img = io.imread(r"C:\Users\marun\Downloads\dog.jpg")

# Convert RGB to HSV
hsv = color.rgb2hsv(img)

# Display
plt.imshow(hsv)
plt.title("HSV Image")
plt.axis("off")
plt.show()
```
