```python
#rgb to xyz open cv code 
import cv2
import matplotlib.pyplot as plt

# Read image
img = cv2.imread(r"C:\Users\marun\Downloads\ccat.gif")

# Convert BGR to RGB
rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Convert RGB to XYZ
xyz = cv2.cvtColor(rgb, cv2.COLOR_RGB2XYZ)

# Display Original and XYZ
plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
plt.imshow(rgb)
plt.title("Original RGB Image")
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(xyz)
plt.title("XYZ Image")
plt.axis('off')

plt.show()

print("RGB Shape:", rgb.shape)
print("XYZ Shape:", xyz.shape)
```

```python
#scikit code 
from skimage import io, color
import matplotlib.pyplot as plt

# Read image
img = io.imread(r"C:\Users\marun\Downloads\dog.jpg")

# Convert RGB to XYZ
xyz = color.rgb2xyz(img)

# Display Original and XYZ
plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
plt.imshow(img)
plt.title("Original RGB Image")
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(xyz)
plt.title("XYZ Image")
plt.axis('off')

plt.show()

print("RGB Shape:", img.shape)
print("XYZ Shape:", xyz.shape)
```
