```python
import cv2
import os
import numpy as np


# Detect face function
def detect_face(img):

    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

    face_cascade = cv2.CascadeClassifier(
        cv2.data.haarcascades +
        "haarcascade_frontalface_default.xml"
    )

    faces = face_cascade.detectMultiScale(
        gray,
        scaleFactor=1.2,
        minNeighbors=5
    )

    if len(faces) == 0:
        return None, None

    (x,y,w,h) = faces[0]

    return gray[y:y+h, x:x+w], faces[0]



# Training data function
def prepare_training_data(folder):

    faces = []
    labels = []

    persons = os.listdir(folder)


    for person in persons:

        if not person.startswith("s"):
            continue


        label = int(person.replace("s",""))

        path = folder + "/" + person


        images = os.listdir(path)


        for image_name in images:


            image_path = path + "/" + image_name

            img = cv2.imread(image_path)


            face, rect = detect_face(img)


            if face is not None:

                faces.append(face)
                labels.append(label)


    return faces, labels



# Load training images

faces, labels = prepare_training_data(
    "training_data"
)



# Create recognizer

recognizer = cv2.face.LBPHFaceRecognizer_create()



# Train model

recognizer.train(
    faces,
    np.array(labels)
)



# Test image

test_img = cv2.imread(r"C:\Users\marun\Downloads\face1.jpg")



face, rect = detect_face(test_img)



# Predict person

label, confidence = recognizer.predict(face)



# Names

names = [
    "",
    "Sachin",
    "Ranveer"
]


name = names[label]



# Draw rectangle

(x,y,w,h) = rect


cv2.rectangle(
    test_img,
    (x,y),
    (x+w,y+h),
    (0,255,0),
    2
)


# Display name

cv2.putText(
    test_img,
    name,
    (x,y-10),
    cv2.FONT_HERSHEY_SIMPLEX,
    1,
    (0,255,0),
    2
)



cv2.imshow(
    "Face Recognition",
    test_img
)


cv2.waitKey(0)
cv2.destroyAllWindows()
```
