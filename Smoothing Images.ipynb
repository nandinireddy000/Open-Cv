```python
import cv2
import matplotlib.pyplot as plt

# Read image
img = cv2.imread(r"C:\Users\marun\Downloads\cat1.jpg")

# Convert BGR to RGB
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Apply smoothing filters
average_blur = cv2.blur(img, (9, 9))
gaussian_blur = cv2.GaussianBlur(img, (9, 9), 0)
median_blur = cv2.medianBlur(img, 9)
bilateral_blur = cv2.bilateralFilter(img, 9, 75, 75)

# Convert outputs to RGB
average_blur = cv2.cvtColor(average_blur, cv2.COLOR_BGR2RGB)
gaussian_blur = cv2.cvtColor(gaussian_blur, cv2.COLOR_BGR2RGB)
median_blur = cv2.cvtColor(median_blur, cv2.COLOR_BGR2RGB)
bilateral_blur = cv2.cvtColor(bilateral_blur, cv2.COLOR_BGR2RGB)

# Display images
plt.figure(figsize=(12, 8))

plt.subplot(2, 3, 1)
plt.imshow(img_rgb)
plt.title("Original")
plt.axis("off")

plt.subplot(2, 3, 2)
plt.imshow(average_blur)
plt.title("Average Blur")
plt.axis("off")

plt.subplot(2, 3, 3)
plt.imshow(gaussian_blur)
plt.title("Gaussian Blur")
plt.axis("off")

plt.subplot(2, 3, 4)
plt.imshow(median_blur)
plt.title("Median Blur")
plt.axis("off")

plt.subplot(2, 3, 5)
plt.imshow(bilateral_blur)
plt.title("Bilateral Filter")
plt.axis("off")

plt.tight_layout()
plt.show()
```
