```python
import cv2
import matplotlib.pyplot as plt

image = cv2.imread(r"C:\Users\marun\Downloads\palm.jpg")

# Convert BGR to RGB for displaying
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

plt.imshow(image_rgb)
plt.title("Original Palm")
plt.axis("off")
plt.show()


gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)


edges = cv2.Canny(gray,40,55,apertureSize=3)


plt.imshow(edges, cmap="gray")
plt.title("Palm Lines")
plt.axis("off")
plt.show()


edges = cv2.bitwise_not(edges)

cv2.imwrite("palmlines.jpg", edges)


palmlines = cv2.imread("palmlines.jpg")

img = cv2.addWeighted(palmlines,0.3,image,0.7,0)

# Convert final image for display
final = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

plt.imshow(final)
plt.title("Final Palm Line Detection")
plt.axis("off")
plt.show()
```
