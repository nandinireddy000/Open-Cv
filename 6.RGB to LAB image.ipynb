```python
import cv2
import matplotlib.pyplot as plt

# Read image
img = cv2.imread(r"C:\Users\marun\Downloads\ccat.gif")

# Convert BGR to RGB
rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Convert RGB to LAB
lab = cv2.cvtColor(rgb, cv2.COLOR_RGB2LAB)

# Display Original and LAB image
plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
plt.imshow(rgb)
plt.title("Original RGB Image")
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(lab)
plt.title("LAB Image")
plt.axis("off")
print("Original RGB Shape:", rgb.shape)
print("LAB Shape:", lab.shape)
plt.show()
```
