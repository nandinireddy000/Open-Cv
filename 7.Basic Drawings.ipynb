```python
#Drawing a line
from skimage import io, draw
import matplotlib.pyplot as plt

img = io.imread(r"C:\Users\marun\Downloads\dog.jpg")

h, w = img.shape[:2]

x, y = draw.line(0, 0, min(h-1, w-1), min(h-1, w-1))

img[x, y] = 0

plt.imshow(img)
plt.axis('off')
plt.show()
```

```python
#Drawing a rectangle
from skimage import io, draw
import matplotlib.pyplot as plt

img = io.imread(r"C:\Users\marun\Downloads\dog.jpg")

# Rectangle start point (row, col)
start = (50, 50)

# Rectangle end point (row, col)
end = (200, 300)

rr, cc = draw.rectangle_perimeter(start=start, end=end)

img[rr, cc] = [255, 0, 0]  # Red rectangle

plt.imshow(img)
plt.axis('off')
plt.show()
```

```python
#Drawing a filled rectangle
from skimage import io, draw
import matplotlib.pyplot as plt

img = io.imread(r"C:\Users\marun\Downloads\dog.jpg")

rr, cc = draw.rectangle(start=(50, 50), end=(200, 300))

img[rr, cc] = [0, 255, 0]  # Green filled rectangle

plt.imshow(img)
plt.axis('off')
plt.show()
```

```python
#Drawing a circle
from skimage import io, draw
import matplotlib.pyplot as plt

img = io.imread(r"C:\Users\marun\Downloads\dog.jpg")

rr, cc = draw.circle_perimeter(250, 250, 100)

img[rr, cc] = [255, 0, 0]  # Red circle

plt.imshow(img)
plt.axis('off')
plt.show()
```

```python
#Drawing a filled circle
from skimage import io, draw
import matplotlib.pyplot as plt

img = io.imread(r"C:\Users\marun\Downloads\dog.jpg")

rr, cc = draw.disk((250, 250), 100)

img[rr, cc] = [0, 0, 255]  # Blue filled circle

plt.imshow(img)
plt.axis('off')
plt.show()
```
