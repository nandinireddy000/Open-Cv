```python
import cv2
import matplotlib.pyplot as plt

# Read image
img = cv2.imread(r"C:\Users\marun\Downloads\img1.webp")

# Convert BGR to RGB
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)


# 1. Resize Image
resize_img = cv2.resize(img, (300,300))
resize_rgb = cv2.cvtColor(resize_img, cv2.COLOR_BGR2RGB)


# 2. Rotate Image
(h, w) = img.shape[:2]

matrix = cv2.getRotationMatrix2D((w//2, h//2), 45, 1)

rotate_img = cv2.warpAffine(img, matrix, (w,h))
rotate_rgb = cv2.cvtColor(rotate_img, cv2.COLOR_BGR2RGB)


# 3. Flip Image
flip_img = cv2.flip(img, 1)   # 1 = horizontal flip
flip_rgb = cv2.cvtColor(flip_img, cv2.COLOR_BGR2RGB)


# 4. Crop Image
crop_img = img[50:300, 50:300]
crop_rgb = cv2.cvtColor(crop_img, cv2.COLOR_BGR2RGB)


# Display output

plt.figure(figsize=(12,8))

plt.subplot(2,3,1)
plt.imshow(img_rgb)
plt.title("Original Image")
plt.axis("off")

plt.subplot(2,3,2)
plt.imshow(resize_rgb)
plt.title("Resized Image")
plt.axis("off")

plt.subplot(2,3,3)
plt.imshow(rotate_rgb)
plt.title("Rotated Image")
plt.axis("off")

plt.subplot(2,3,4)
plt.imshow(flip_rgb)
plt.title("Flipped Image")
plt.axis("off")

plt.subplot(2,3,5)
plt.imshow(crop_rgb)
plt.title("Cropped Image")
plt.axis("off")

plt.show()
```
