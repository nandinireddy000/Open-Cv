```python
import cv2
import matplotlib.pyplot as plt

# Read image in grayscale
img = cv2.imread(r"C:\Users\marun\Downloads\cat1.jpg", 0)

# Apply different thresholding techniques
_, thresh_binary = cv2.threshold(img, 127, 255, cv2.THRESH_BINARY)
_, thresh_binary_inv = cv2.threshold(img, 127, 255, cv2.THRESH_BINARY_INV)
_, thresh_trunc = cv2.threshold(img, 127, 255, cv2.THRESH_TRUNC)
_, thresh_tozero = cv2.threshold(img, 127, 255, cv2.THRESH_TOZERO)

# Display results
plt.figure(figsize=(12,8))

plt.subplot(2,3,1)
plt.imshow(img, cmap='gray')
plt.title("Original")
plt.axis("off")

plt.subplot(2,3,2)
plt.imshow(thresh_binary, cmap='gray')
plt.title("Binary")
plt.axis("off")

plt.subplot(2,3,3)
plt.imshow(thresh_binary_inv, cmap='gray')
plt.title("Binary Inverse")
plt.axis("off")

plt.subplot(2,3,4)
plt.imshow(thresh_trunc, cmap='gray')
plt.title("Truncate")
plt.axis("off")

plt.subplot(2,3,5)
plt.imshow(thresh_tozero, cmap='gray')
plt.title("To Zero")
plt.axis("off")

plt.tight_layout()
plt.show() 
```
